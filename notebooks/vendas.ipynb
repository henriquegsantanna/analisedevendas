# Planejamento Inicial

-   Dropar as colunas 'Row ID', 'Ship Date', 'Ship Mode', 'City', 'Postal Code' e 'Sub-Category' (não vou usá-las)
-   Organizar as colunas 'Sales' e 'Profit' para somente 2 casas decimais
-   Verificar valores nulos, duplicados e faltantes
-   Usar df.describe para ver se tem algum dado estranho/inconsistente
-   Coluna 'Order Date' está como objeto, transformar em datetime
-   Colunas 'Customer Name', 'Segment', 'Country', 'State', 'Region' e 'Category' estão como objeto, transformar em str

```python
df.drop(['Row ID', 'Ship Date', 'Ship Mode', 'City', 'Postal Code', 'Sub-Category'], axis=1, inplace=True) #axis=0 aponta para linhas e axis=1 aponta para colunas / inplace=true altera diretamente no df original
```

```python
df = df.round({'Sales': 2, 'Profit': 2}) #arredonda pra 2 casas decimais as duas colunas
```

```python
df.info() #verificando valores nulos
```

```python
df.duplicated().sum()
```

```python
duplicadas = df[df.duplicated(keep=False)] #cria uma lista com as linhas duplicadas
print(duplicadas)
```

```python
df.drop(3406, inplace=True) #remove a linha de indice 3406 (a linha duplicada)
```

'Sales e 'Profit' estão com um desvio-padrão muito alto (as colunas possuem valores muito altos e muito baixos):
-   'Sales': média de 229.85 com desvio-padrão de 623.27
-   'Profit': média de 28.66 de lucro com desvio-padrão de 234.27

Os valores de mínimo e máximo das colunas 'Sales e 'Profit' confirmam isso:
-   'Sales': 0.44 mínimo / 22638.48 máximo
-   'Profit': -6599.98 mínimo / 8399.98 máximo

Analisando linhas inconsistentes:

```python
colunas_filtradas = ['Product Name', 'Sales', 'Quantity', 'Discount', 'Profit'] #colunas que serão exibidas no print

print(
    df.sort_values(by='Sales', ascending=False) #ordena pela coluna 'Sales' do maior pro menor e printa os 10 maiores
    .head(10)[colunas_filtradas]
)
```

-   Aparentemente normal, o produto era caro e foram compradas 6 unidades, profit ficou negativo acredito que por conta do desconto de 50%.
-   Os dados dos 10 produtos mais caros não apresentaram ampla discrepância, profit também não mudou muito de acordo com o valor e quantidade do produto.

```python
colunas_filtradas = ['Product Name', 'Sales', 'Quantity', 'Discount', 'Profit'] #colunas que serão exibidas no print

print(
    df.sort_values(by='Profit', ascending=False) #ordena pela coluna 'Profit' do maior pro menor e printa os 10 maiores
    .head(10)[colunas_filtradas]
)
```

-   Normal, produtos caros vendidos em quantidade e sem desconto, consequentemente o lucro foi maior.

```python
colunas_filtradas = ['Product Name', 'Sales', 'Quantity', 'Discount', 'Profit'] #colunas que serão exibidas no print

print(
    df.sort_values(by='Profit', ascending=True) #ordena pela coluna 'Profit' do menor pro maior e printa os 10 maiores
    .head(10)[colunas_filtradas]
)
```

-   Normal, descontos muito altos resultaram em profits negativos

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y') #transformando a coluna 'Order Date' em tipo datetime
```
